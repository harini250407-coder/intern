OUTLIERS DETECTION AND HANDLING

```python
import seaborn as sns
sns.boxplot(df1['Age'])
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1-1.5*IQR
upper_bound = Q3+1.5*IQR

outliers = df1[(df1['Age']<lower_bound) | (df1['Age']>upper_bound)]
outliers
```

```python
from scipy.stats import zscore
z_scores = zscore(df1[['Age']])
df1[np.abs(z_scores) > 2.5]
```

```python
from sklearn.ensemble import IsolationForest
iso = IsolationForest(contamination = 0.1)
df1['outliers'] = iso.fit_predict(df1[['Age','Salary']])
df1[df1['outliers'] == -1]
```

```python
from sklearn.ensemble import IsolationForest
iso = IsolationForest(contamination = 0.5)
df['outlier'] = iso.fit_predict(df[['Age','Salary']])
df[df['outlier'] == -1]
```

```python
from scipy.stats.mstats import winsorize
df['Age'] = winsorize(df['Age'],limits = [0.05,0.05])
df.head()
```

```python
#log transform
df['Age'] = np.log(df['Age'])
df.head()
```

```python
#POWER TRANSFORMER
from sklearn.preprocessing import PowerTransformer
pt = PowerTransformer(method = 'box-cox')
df['Salary'] = pt.fit_transform(df[['Salary']])
df.head()
```

```python
from sklearn.preprocessing import PowerTransformer
pt = PowerTransformer(method = 'yeo-johnson')
df1['Salary'] = pt.fit_transform(df1[['Salary']])
df1.head()
```

```python
#minmax and standard scalar
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df['Age'] = scaler.fit_transform(df[['Age']])
df.head()
```
