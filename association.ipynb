```python
transactions=[]
for i in range(0,len(d)):
  transactions.append([str(d.values[i,j])for j in range (0,20)
    if str(d.values[i,j])!='NaN'])
```

```python
!pip install apyori
```

```python
from apyori import apriori
rules=apriori(transactions=transactions,min_support=0.003,min_confidence=0.4,min_left=3,min_length=2,max_length=2)
results=list(rules)
results
```

```python
def inspect(results):
    lhs         = [tuple(rule[2][0][0])[0] for rule in results if 'nan' not in rule[0]]
    rhs         = [tuple(rule[2][0][1])[0] for rule in results if 'nan' not in rule[0]]
    supports    = [rule[1] for rule in results if 'nan' not in rule[0]]
    confidences = [rule[2][0][2] for rule in results if 'nan' not in rule[0]]
    lifts       = [rule[2][0][3] for rule in results if 'nan' not in rule[0]]
    return list(zip(lhs, rhs, supports, confidences, lifts))
results_df = pd.DataFrame(inspect(results), columns = ['Left Hand Side', 'Right Hand Side', 'Support', 'Confidence', 'Lift'])
display(results_df)
```
