```python
import pandas as pd
from pyECLAT import ECLAT
```

```python
transactions = []
for i in range(len(d)):
    transactions.append(
        [str(d.values[i, j])
         for j in range(len(d.columns))
         if str(d.values[i, j]) != 'nan']
    )
```

```python
df = pd.DataFrame(transactions)
```

```python
eclat = ECLAT(data=df)
```

```python
indices, support = eclat.fit(min_support=0.03,
                             min_combination=1,
                             max_combination=3)
```

```python
result = pd.DataFrame({
    'Itemset': support.keys(),
    'Support': support.values()
})
```

```python
result = result.sort_values(by='Support', ascending=False)
```
